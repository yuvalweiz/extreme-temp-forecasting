Daily AGG

```python
import os
import pandas as pd
import numpy as np

# File paths
hourly_data_path = r'C:\תואר שני\מחקר נגה\נתונים שעתיים\new_isr_gnd_obs_web.csv'
station_table_path = r'new_stn_table_web'
output_dir_hourly = r'C:\תואר שני\מחקר נגה\נתונים שעתיים\Stations Hourly Data 03_08_2025'
output_dir_daily = r'C:\תואר שני\מחקר נגה\נתונים שעתיים\Stations  Daily Data 03_08_2025'

# Ensure output directory exists
os.makedirs(output_dir_daily, exist_ok=True)

def calculate_heat_stress(row):
    humidity = row['hmd_rlt']
    temperature = row['tmp_air_dry']
    e_term = np.exp(((18.678 - temperature) / 234.5) * (temperature / (257.14 + temperature)))
    heat_stress = (((6.1121 * humidity) / 100) * e_term) * 0.348 + temperature - 4.25
    return heat_stress

# Process each hourly station file and aggregate to daily data
for file in os.listdir(output_dir_hourly):
    if file.endswith("_hourly_data.csv"):
        station_name = file.replace("_hourly_data.csv", "")
        print("processing "+ station_name)
        file_path = os.path.join(output_dir_hourly, file)
        data = pd.read_csv(file_path)

        # Calculate heat stress for each row
        data['heat_stress'] = data.apply(calculate_heat_stress, axis=1)

        # Perform daily aggregation for mean values
        daily_means = data.groupby(['year', 'month', 'day']).agg({
            'prs_stn': 'mean',
            'prs_sea_lvl': 'mean',
            'prs_lvl_hgt': 'mean',
            'tmp_air_dry': 'mean',
            'tmp_air_wet': 'mean',
            'tmp_dew_pnt': 'mean',
            'hmd_rlt': 'mean',
            'wind_dir': 'mean',
            'wind_spd': 'mean',
            'cld_ttl_cvr': 'mean',
            'cld_low_cvr': 'mean',
            'cld_low_hgt': 'mean',
            'cld_low_type': 'mean',
            'cld_med_type': 'mean',
            'cld_high_type': 'mean',
            'weather_crr': 'mean',
            'weather_past_1': 'mean',
            'yearly_rain': 'mean',
        }).reset_index()

        # Calculate max and min values for specific columns
        daily_max_min = data.groupby(['year', 'month', 'day']).agg({
            'tmp_air_dry': ['max', 'min'],
            'tmp_air_wet': ['max', 'min'],
            'hmd_rlt': ['max', 'min'],
            'heat_stress': ['max', 'min']
        }).reset_index()

        # Flatten the multi-level columns for max/min DataFrame
        daily_max_min.columns = ['year', 'month', 'day', 
                                 'max_dry_temp', 'min_dry_temp', 
                                 'max_wet_temp', 'min_wet_temp', 
                                 'max_rel_hum', 'min_rel_hum', 
                                 'max_heat_stress', 'min_heat_stress']

        # Merge the daily means and max/min data
        daily_data = pd.merge(daily_means, daily_max_min, on=['year', 'month', 'day'])

        # Save daily aggregated data
        output_path = os.path.join(output_dir_daily, f"{station_name}_daily_data.csv")
        daily_data.to_csv(output_path, index=False)

print(f"Daily aggregated data has been saved to {output_dir_daily}")
```

Merge Daily Max/Min Values

```python
import os
import pandas as pd
from tqdm import tqdm

# === CONFIG ===
stations_folder = "C:\תואר שני\מחקר נגה\נתונים שעתיים\Stations Daily Data 03_08_2025"  # 📁 Replace with your actual folder path
stn_table_path = "new_stn_table_web.csv"
high_precision_data_path = "new_isr_daily_data_web.csv"

# === Step 1: Load station mapping ===
stn_table_df = pd.read_csv(stn_table_path)
stn_dict = {
    row['stn_num']: row['stn_name'].replace(" ", "_") + ".csv"
    for _, row in stn_table_df.iterrows()
}

# === Step 2: Load and prep high-precision data ===
high_precision_df = pd.read_csv(high_precision_data_path)
high_precision_df['Date'] = pd.to_datetime(high_precision_df['time_obs']).dt.strftime('%Y-%m-%d')

# Keep only necessary columns
high_precision_df = high_precision_df[['stn_num', 'Date', 'tmp_air_max', 'tmp_air_min']]

# === Step 3: Group by stn_num and update in bulk ===
for stn_num, group_df in tqdm(high_precision_df.groupby('stn_num'), desc="Merging station updates"):
    if stn_num not in stn_dict:
        continue

    station_filename = stn_dict[stn_num]
    station_path = os.path.join(stations_folder, station_filename)

    if not os.path.exists(station_path):
        continue

    try:
        station_df = pd.read_csv(station_path)
    except Exception as e:
        print(f"Error loading {station_filename}: {e}")
        continue

    if 'Date' not in station_df.columns:
        continue

    # Merge on Date
    merged = pd.merge(
        station_df,
        group_df,
        on='Date',
        how='left',
        suffixes=('', '_new')
    )

    # Replace values if new ones exist
    for col_old, col_new in [('max_dry_temp', 'tmp_air_max'), ('min_dry_temp', 'tmp_air_min')]:
        if col_new in merged.columns:
            merged[col_old] = merged[col_new].combine_first(merged[col_old])

    # Drop temp columns
    merged = merged.drop(columns=['tmp_air_max', 'tmp_air_min'], errors='ignore')

    # Save back
    merged.to_csv(station_path, index=False)

print("✅ All stations updated via fast joins.")
```
